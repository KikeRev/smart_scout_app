```python
import pandas as pd
```

```python
temp_1 = "data/all_players_cleaned.csv"
hist   = "data/historical_players_raw.csv"
```

```python
team = "Real Madrid"
```

```python
df1 = pd.read_csv(temp_1)
df2 = pd.read_csv(hist)
```

```python
df1["Season"] = "2024-25"
```

```python
df2.drop("matches", axis=1, inplace=True)
```

```python
df2 = df2[df1.columns]
```

```python
df1.shape, df2.shape
```

```python
df1.head(1)
```

```python
df2.head(1)
```

```python
df_final = pd.concat([df1, df2], axis=0)
```

```python
df_final.shape
```

```python
df2.player.nunique(), df1.player.nunique()
```

```python
len(set(df1.player.unique()).difference(set(df2.player.unique())))
```

```python
df_final["minutes"] = df_final["minutes"].fillna("0")
df_final["minutes.1"] = df_final["minutes.1"].fillna("0") 
```

```python
df_final["minutes"] = df_final["minutes"].apply(lambda x: float(x.replace(",","")))
df_final["minutes.1"] = df_final["minutes.1"].apply(lambda x: float(str(x).replace(",",".")))
```

```python
df_final["minutes.1"].describe()
```

```python
df_final.info(verbose=True, show_counts=True)
```

```python
cat_cols = ["player", "nationality", "position","age", "Team", "League", "Team_Logo", "Season"]
num_cols = ["player"] + [c for c in df_final.columns if c not in cat_cols]
```

```python
df_cat = df_final[cat_cols]
```

```python
df_num = df_final[num_cols]
```

```python
 df_cat.sort_values(["player", "Season"], ascending=[False, False]).groupby("player").first()
```

```python
df_cat_agg = df_cat.sort_values(["player", "Season"], ascending=[False, False]).groupby("player").first().reset_index()
```

```python
assert df_cat_agg.shape[0] == df_final["player"].nunique()
```

```python
df_cat_agg.Season.value_counts()
```

```python
df_cat_agg["player_status"] = df_cat_agg["Season"].apply(lambda x: "active" if x == "2024-25" else "retired or inactive")
```

```python
df_num[df_num.minutes == 0].shape, df_num.shape
```

```python
df_final[df_final.minutes == 0]
```

```python
df_num_agg = df_num[df_num.minutes > 0].groupby("player").mean().reset_index()
```

```python
df_num_agg.shape
```

```python
df_final_agg = df_cat_agg.merge(df_num_agg, on = "player", how="left")
```

```python
df_final_agg.shape
```

```python
df_final_agg.info(verbose=True, show_counts=True)
```

```python
cols = list(df1.columns) + ["player_status"]
```

```python
df_final_agg = df_final_agg[cols]
```

```python
df_final_agg.info(verbose=True, show_counts=True)
```

```python
df_final_agg.to_csv("data/all_players_plus_historic_data.csv")
```
